```python
try:
    no=int(no)
except ValueError as e:
    print("error : ",e)
except Exception as e:
    print("Exception : ",e)
```

```python
try:
    if no.isdigit() is False:
        raise ValueError("Provided input is string no a valid number")
except ValueError as e:
    print("error : ",e)
except Exception as e:
    print("Exception : ",e)
```
