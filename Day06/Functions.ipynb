```python
#def function_name(inputParameter1,inputParameter1):
    #logic

def calculator(no1,no2,operator):
    match operator:
        case '+':
            print(no1+no2)
    
        case '-':
            print(no1-no2)
```

```python
calculator(145,145,'+')
```

```python
calculator(145,145,'-')
```

```python
#default parameters
def add(no1,no2=10):
    print("Addition: ",no1+no2)
```

```python
#Keyword parameters
```

```python
def fname(first_name,last_name):
    print(first_name," ",last_name)
```

```python
fname(last_name='XYZ',first_name='ABC')
```

```python
#Positional parameters
```

```python
fname("ABC","XYZ")
```

```python
#Arbitary Arguments
def myFun(*args, **kwargs):
   print(args)
   print(kwargs)
   print(args[0])
   print(kwargs['first'])
   
myFun('Hi', 'Welcome', first='Python', last='Functions',second='parameters')
```
