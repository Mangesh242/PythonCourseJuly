```python
# Create a set
numbers = {1, 2, 3, 4}

#Read:
print(numbers)         # {1, 2, 3, 4}
print(2 in numbers)    # True

#Update:
numbers.add(5)         # add element
print(numbers)         # {1, 2, 3, 4, 5}

numbers.update([6, 7]) # add multiple elements
print(numbers)         # {1, 2, 3, 4, 5, 6, 7}

#Delete:
numbers.remove(3)      # remove element (error if not found)
print(numbers)         # {1, 2, 4, 5, 6, 7}
numbers.discard(10)    # safe remove (no error if not found)
numbers.pop()          # removes a random element
print(numbers)
numbers.clear()        # remove all elements
print(numbers)         # set()
```
