```python
class Student:
    def __init__(self, name, marks): #Constuctors : Object  banvnara
        self.name = name
        self.marks = marks
    def display(self):  #Behavior
        print(f"Name: {self.name}, Marks: {self.marks}")
```

```python
obj= Student("Rahul",95)
```

```python
print(type(obj))
```

```python
obj.display()
```

```python
#Inheritance 
class Animal:                         #parent
    def speak(self):
        print("Animal speaks")
class Dog(Animal):		#Children Class
    def ear(self):
        print("Dog eats")
d = Dog()
```

```python
d.speak()
```

```python
class Bank:
    def __init__(self,bank_balance,balance):
        self.__bank_balance=bank_balance
        self.balance=balance
    def getPrivateBalance(self):
        return self.__bank_balance
    def setPriBalance(self,balance):
        self.__bank_balance=balance
```

```python
b = Bank(1000,10)
```

```python
#public variable we could access directly
print(b.balance)
```

```python
b.getPrivateBalance()
```
